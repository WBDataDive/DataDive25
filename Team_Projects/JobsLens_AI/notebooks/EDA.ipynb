```python
import pandas as pd
```

```python
join_dataset = pd.read_excel('data/join_dataset.xlsx')
join_dataset.columns = join_dataset.iloc[2]
join_dataset = join_dataset.iloc[3:]
# Normalize column names: lowercase, replace spaces with underscores, remove special characters
join_dataset.columns = (
    join_dataset.columns
        .str.strip()
        .str.lower()
        .str.replace(' ', '_')
        .str.replace(r'[^\w_]', '', regex=True)
)

# data quality flags filter
# join_dataset = join_dataset[join_dataset['percentile_100__max_share_of_flags'].astype(float) < 30]

# filter subsample for 'all' only
join_dataset = join_dataset[join_dataset['subsample'] == 'Urban']

# filter the dataset to only include the latest date for each country
join_dataset['year_of_survey'] = join_dataset['year_of_survey'].astype(int)
latest_date = join_dataset.groupby('country_code', as_index=False)['year_of_survey'].max()
# merge the latest date with the dataset so we can filter the dataset to only include the latest date for each country  
join_dataset = join_dataset.merge(latest_date, on=['country_code', 'year_of_survey'], how='inner')

join_dataset.head()
```

```python
hai_data = pd.read_csv('data/hai_full_database.csv')

hai_data.columns = hai_data.columns.str.strip().str.lower().str.replace(' ', '_').str.replace(r'[^\w_]', '', regex=True)

hai_data.head()
```
